## Dicionário de dados

Há cinco tabelas no conjunto de dados, e você vai precisar usar todas elas para pré-processar seus dados e fazer AED. Abaixo está um dicionário que lista as colunas de cada tabela e descreve os dados contidos nelas.

- `instacart_orders.csv`: cada linha corresponde a um pedido no aplicativo da Instacart
    - `'order_id'`: é o número que identifica cada pedido de forma exclusiva
    - `'user_id'`: é o número de identificação exclusivo da conta de cada cliente
    - `'order_number'`: é o número de vezes que o cliente fez um pedido
    - `'order_dow'`: é o dia da semana em que o pedido foi feito (0 é domingo)
    - `'order_hour_of_day'`: é a hora do dia em que o pedido foi feito
    - `'days_since_prior_order'`: é o número de dias desde que o cliente fez seu pedido anterior




- `products.csv`: cada linha corresponde a um produto exclusivo que os clientes podem comprar
    - `'product_id'`: é o número de identificação unívoco de cada produto
    - `'product_name'`: é o nome do produto
    - `'aisle_id'`: é o número de identificação exclusivo de cada categoria de corredor do supermercado
    - `'department_id'`: é o número de identificação exclusivo de cada categoria de departamento do supermercado




-	`order_products.csv`: cada linha corresponde a um item incluído em um pedido
    -	`'order_id'`: é o número que identifica cada pedido de forma exclusiva
    -	`'product_id'`: é o número de identificação exclusivo de cada produto
    -	`'add_to_cart_order'`: é a ordem sequencial em que cada item foi colocado no carrinho
    -	`'reordered'`: 0 se o cliente nunca comprou o produto antes, 1 se já o comprou




-	`aisles.csv`
    -	`'aisle_id'`: é o número de identificação exclusivo de cada categoria de corredor do supermercado
    -	`'aisle'`: é o nome do corredor



-	`departments.csv`
    -	`'department_id'`: é o número de identificação exclusivo de cada categoria de departamento do supermercado
    -	`'department'`: é o nome do departamento

```python


#Carregar os dados 

import pandas as pd
from pathlib import Path

def read_instacart_csv(filename):
    for p in [Path(filename), Path("datasets")/filename, Path("/datasets")/filename]:
        if p.exists():
            print(f"✅ Lendo: {p}")
            # os arquivos do projeto usam ponto e vírgula
            return pd.read_csv(p, sep=';', engine='python')
    raise FileNotFoundError(f"Arquivo '{filename}' não encontrado em '.', './datasets' nem '/datasets'.")

orders         = read_instacart_csv('instacart_orders.csv')
products       = read_instacart_csv('products.csv')
order_products = read_instacart_csv('order_products.csv')
aisles         = read_instacart_csv('aisles.csv')
departments    = read_instacart_csv('departments.csv')

print('orders:', orders.shape)
print('products:', products.shape)
print('order_products:', order_products.shape)
print('aisles:', aisles.shape)
print('departments:', departments.shape)


```

```python
# Imprimir as informações sobre o DataFrame principal (orders)

print('Dimensões:', orders.shape)
print('\nInformações gerais:')
orders.info()

print('\nValores nulos por coluna:')
print(orders.isna().sum())

print('\nPrimeiras linhas:')
display(orders.head())

 # imprima as informações sobre o DataFrame
```

```python
# Limpar os dados

# Remover duplicados
orders.drop_duplicates(inplace=True)

# Tratar valores ausentes
orders['days_since_prior_order'].fillna(orders['days_since_prior_order'].median(), inplace=True)

# Verificar se ficou tudo limpo
print('Valores ausentes:')
print(orders.isna().sum())

print('\nDimensões após limpeza:')
print(orders.shape)
```

```python
# Unir os datasets

data = orders.merge(order_products, on='order_id') \
             .merge(products, on='product_id') \
             .merge(aisles, on='aisle_id') \
             .merge(departments, on='department_id')

# Imprimir informações sobre o DataFrame unificado
print('Dimensões do DataFrame:', data.shape)
print('\nInformações gerais:')
data.info()
```

```python
# Analisar os dados

print('Dimensões do DataFrame:')
print(data.shape)

print('\nInformações gerais:')
data.info()

print('\nEstatísticas descritivas:')
print(data.describe())

print('\nValores ausentes por coluna:')
print(data.isna().sum())

 # imprima as informações sobre o DataFrame
```

### [C3] Para cada produto, qual parcela de todos os pedidos dele são repetidos?

```python
# Contar total de pedidos por produto
total_por_produto = order_products.groupby('product_id').size().reset_index(name='total_pedidos')

# Contar só os pedidos repetidos por produto
repetidos_por_produto = order_products[order_products['reordered'] == 1] \
    .groupby('product_id').size().reset_index(name='pedidos_repetidos')

# Juntar as duas contagens
proporcao = total_por_produto.merge(repetidos_por_produto, on='product_id', how='left')

# Substituir NaN por 0 (produtos nunca reordenados)
proporcao['pedidos_repetidos'] = proporcao['pedidos_repetidos'].fillna(0)

# Calcular a proporção
proporcao['proporcao_repetida'] = proporcao['pedidos_repetidos'] / proporcao['total_pedidos']

# Adicionar o nome do produto
proporcao = proporcao.merge(products[['product_id', 'product_name']], on='product_id', how='left')

# Exibir os 10 primeiros como exemplo
proporcao.sort_values(by='proporcao_repetida', ascending=False).head(10)
```

Escreva suas conclusões aqui

A maioria dos produtos tem uma parcela considerável de pedidos repetidos, o que sugere que os clientes tendem a comprar novamente os mesmos itens. 
Os produtos com maior proporção de repetição geralmente são itens essenciais ou de consumo frequente. 
Esta métrica é útil para prever reposições de stock e identificar itens de fidelidade. Produtos com baixa taxa de repetição podem indicar compras experimentais ou sazonais.

### [C4] Para cada cliente, qual proporção de todos os seus pedidos são repetidos?

```python
# Converter tipos
order_products['product_id'] = pd.to_numeric(order_products['product_id'], errors='coerce')
order_products['reordered'] = pd.to_numeric(order_products['reordered'], errors='coerce')

# Garantir tipos corretos
orders['order_id'] = pd.to_numeric(orders['order_id'], errors='coerce')
orders['user_id'] = pd.to_numeric(orders['user_id'], errors='coerce')

# Juntar os dois datasets para associar pedidos a clientes
merged = order_products.merge(orders[['order_id', 'user_id']], on='order_id', how='left')

# Calcular a proporção de pedidos repetidos por cliente
customer_reorders = (
    merged.groupby('user_id')['reordered']
    .mean()
    .reset_index()
    .rename(columns={'reordered': 'proportion_reordered'})
)

# Mostrar as primeiras linhas
customer_reorders.head()
```
